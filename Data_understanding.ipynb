```python
#Importing libraries and dataset


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

train=pd.read_csv("E:/Laptop/mini project 3/Data/train.csv")
test=pd.read_csv("E:/Laptop/mini project 3/Data/test.csv")

print("Train Shape:",train.shape)
print("Test Shape:",test.shape)

train.head()
```

```python
#Getting the info of the dataset

train.info()
```

```python
# Summary statistics of the dataset

train.describe()
```

```python
#Checking for null values

missing=train.isnull().sum()

missing=missing[missing>0].sort_values(ascending=False)

print(missing)
```

```python
#visualize distribution of missing values

plt.figure(figsize=(12,6))

missing.plot(kind='bar')

plt.xticks(rotation=75)

plt.title("Missing Values")

plt.show()
```

```python
#Premium distribution based on age scatter plot

plt.figure(figsize=(12,6))
sns.scatterplot(x='Age',y='Premium Amount',data=train)
plt.title("Premium Distribution based on Age")
plt.show()
```

```python
#checking premium distribution based on occupation box plot

plt.figure(figsize=(12,6))
sns.boxplot(data=train, x='Occupation', y='Premium Amount')
plt.xticks(rotation=45)
plt.show()
```

```python
#premium distribution based on Annual Income scatter plot
plt.figure(figsize=(12,6))
sns.scatterplot(x='Annual Income',y='Premium Amount',data=train)
plt.title("Premium Distribution based on Annual Income")
plt.show()
```

```python
#premium distribution based credit score histogram
plt.figure(figsize=(12,6))
sns.histplot(data=train, x='Credit Score', bins=30, kde=True)
plt.title("Premium Distribution based on Credit Score")
plt.show()
```

```python
#premium distribution based on duration histogram
plt.figure(figsize=(12,6))
sns.histplot(data=train, x='Insurance Duration', bins=30, kde=True)
plt.title("Premium Distribution based on Insurance Duration")
plt.show()
```

```python
#premium distribution based on smoking status using count plot
plt.figure(figsize=(12,6))
sns.countplot(data=train, x='Smoking Status')
plt.title("Premium Distribution based on Smoking Status")
plt.show()
```

```python
#Checking correlation between numerical features
numerical_features=train.select_dtypes(include=[np.number])
correlation=numerical_features.corr()
plt.figure(figsize=(12,8))
sns.heatmap(correlation,annot=True,cmap='coolwarm',linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

```python
# numerical vs categorical features

num_cols=train.select_dtypes(include=np.number).columns

cat_cols=train.select_dtypes(include='object').columns

print("Numerical:",num_cols)

print()

print("Categorical:",cat_cols)
```
